# Dependence architecture ablation

Focused review of the **blockwise empirical benchmark** and the **hierarchical global/local
cross-asset architecture**. It uses the frozen v30 score summary, so the notebook is fast and the
architecture comparison is not obscured by a full data/refit loop.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
ROOT = Path.cwd()
if ROOT.name == "notebooks": ROOT = ROOT.parent
scores = pd.read_csv(ROOT / "results" / "v30_score_summary.csv")
```

```python
direct = scores[scores["propagation"] == "direct_H"].copy()
best_by_arch = direct.groupby(["horizon","architecture"],as_index=False).agg(
    energy_score=("energy_score","min"),
    variogram_score=("variogram_score","min"))
display(best_by_arch)
```

```python
order=["1W","1M","3M","1Y"]
best_by_arch.pivot(index="horizon",columns="architecture",values="energy_score").reindex(order).plot(kind="bar",figsize=(9,5))
plt.title("Direct-H: best Energy Score by dependence architecture")
plt.ylabel("Energy Score (lower is better)")
plt.tight_layout(); plt.show()
```

### Interpretation

The hierarchy is the more structured research architecture, but blockwise resampling remains the hard
benchmark. In the frozen lean adjudication, hierarchical specifications delivered the best direct-H
Energy Score at 1W, 3M and 1Y; blockwise was best at 1M. Dependence gains remain pair-specific rather
than uniform across the entire matrix.
