# Validation and proper scoring rules

Compact evidence notebook for the frozen public snapshot. It reads committed reference results and
reproduces the headline diagnostics without downloading data or refitting the engine.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
ROOT=Path.cwd()
if ROOT.name=="notebooks": ROOT=ROOT.parent
scores=pd.read_csv(ROOT/"results"/"v30_score_summary.csv")
effects=pd.read_csv(ROOT/"results"/"v30_main_effects.csv")
components=pd.read_csv(ROOT/"results"/"v30_variogram_components.csv")
v23=pd.read_csv(ROOT/"results"/"v23_var_es_reference.csv")
display(v23)
```

```python
order=["1W","1M","3M","1Y"]
e=effects[effects["metric"]=="energy_score"].pivot(index="horizon",columns="factor",values="mean_gain").reindex(order)
e.plot(kind="bar",figsize=(9,5))
plt.axhline(0,linewidth=1); plt.title("Factorial main effects on OOS Energy Score")
plt.ylabel("Mean improvement (positive = better)"); plt.tight_layout(); plt.show()
```

```python
h=components[components["factor"]=="hierarchy"].set_index("horizon").reindex(order)
h[["EQ_RATES","EQ_CREDIT","RATES_CREDIT","FX_CROSS"]].plot(kind="bar",figsize=(9,5))
plt.axhline(0,linewidth=1); plt.title("Hierarchy: Variogram component gains")
plt.ylabel("Mean improvement (positive = better)"); plt.tight_layout(); plt.show()
```

### Frozen interpretation

- **Direct-H** remains the default propagation choice.
- **Hierarchy** is the research architecture; **blockwise** remains the permanent benchmark.
- **Duration-conditioned PIT pooling** remains an optional challenger because evidence is horizon-dependent.
- The annual row has low statistical power and is treated as descriptive.
